# 01 Bronze ingest
Loads the three M5 CSVs from the volume into Delta tables without any changes.

```python
RAW = "/Volumes/workspace/m5/raw"
SCHEMA = "workspace.m5"

files = {
    "sales_wide": "sales_train_evaluation.csv",
    "calendar":   "calendar.csv",
    "prices":     "sell_prices.csv",
}

for name, file in files.items():
    df = spark.read.csv(f"{RAW}/{file}", header=True, inferSchema=True)
    df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(f"{SCHEMA}.bronze_{name}")
    print(f"bronze_{name}: {df.count():,} rows, {len(df.columns)} columns")
```

```python
display(spark.table("workspace.m5.bronze_sales_wide").select("id", "item_id", "store_id", "d_1", "d_2", "d_1941").limit(5))
display(spark.table("workspace.m5.bronze_calendar").limit(5))
display(spark.table("workspace.m5.bronze_prices").limit(5))
```
