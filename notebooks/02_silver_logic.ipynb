```python
silver = build_silver(spark)

# Hold back the last 28 days: Part F loads them later with MERGE, to simulate new data arriving
max_date = silver.agg(F.max("date")).first()[0]
cutoff = F.date_sub(F.lit(max_date), 27)

(silver.filter(F.col("date") < cutoff)
       .write.mode("overwrite")
       .option("overwriteSchema", "true")
       .saveAsTable("workspace.m5.silver_sales_daily"))

print("max date in data:", max_date)
```

```python
s = spark.table("workspace.m5.silver_sales_daily")
print(f"rows: {s.count():,}")
print("dates:", s.agg(F.min("date"), F.max("date")).first())
display(s.filter("id = 'FOODS_3_090_CA_3_evaluation'").orderBy("date").limit(10))
```
